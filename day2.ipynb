```python
#init is a constructor 
#show is a instance method
class student:
    def __init__(self,name,place):
        self.name=name
        self.place=place
    def show(self):
        return f'my name is {self.name} and i am form {self.place}.'
p1=student("theju","basavapatna")
p1.show()
```

```python
#init is a constructor 
#show is a instance method
class student:
    def __init__(stu,name,place,usn):
        stu.name=name
        stu.place=place
        stu.usn=usn
    def show(stu):
        return f'usn is {stu.usn} my name is {stu.name} and i am form {stu.place}.'
    def update(stu,place,usn):
        stu.place=place 
        stu.usn=usn
p1=student("theju","basavapatna",18)
p1.show()
```
