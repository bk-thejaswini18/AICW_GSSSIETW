```python
hr=hr.drop(['Floor','Area Locality'],axis=1)
hr.columns
```

```python
#data splitting =100% -80% TRAINING SET AND 20% testing set
x=hr.drop(columns='Rent') #x is independent variable
y=hr['Rent']  #y is dependent variable
```

```python
#train test split function
x_train,x_test,y_train,y_test=train_test_split(x,y,train_size=0.8,random_state=42)
```

```python
x_train.shape,x_test.shape
```

```python
#train
lr=LinearRegression()
```

```python
model=lr.fit(x_train,y_train)
model
```

```python
#predict
y_pred=model.predict(x_test)
y_pred
```

```python
#evaluate the model
mse=mt.mean_squared_error(y_test,y_pred)
mae=mt.mean_absolute_error(y_test,y_pred)
r2=mt.r2_score(y_test,y_pred)
print(f'the mean squared error is {mse}')
print(f'the mean squared error is {mae}')
print(f'the mean squared error is {r2}')
```
